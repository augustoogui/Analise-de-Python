# Exercícios

```python
import pandas as pd
dados = {"Nome": ["Ana", "Bruno", "Carla", "Diego", "Elias"],
         "Idade": [25, 30, 45, 32, 28,],
         "Cidade":["São Paulo", "Guarulhos", "São Paulo", "Campinas", "Campinas"]
         }

df = pd.DataFrame(dados)
print(df)
```

# 2.

```python
# selecionar apenas a coluna "Idade"
idades = df["Idade"]
print(idades)
```

```python
# Selecionar linhas onde a idade é maior que 25
maiores_25 = df[df["Idade"] > 25]

print(maiores_25)
```

# 3.

```python
df["Maior de idade"] = df["Idade"] > 18
print(df)
```
